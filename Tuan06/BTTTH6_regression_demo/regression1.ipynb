```python
train_data = [
    [170, 65],
    [155, 50],
    [150, 45],
    [175, 70],
    [165, 55], 
]
print(train_data)
```

```python
import sympy as sym

a = sym.Symbol('a')
b = sym.Symbol('b')

residual = 0
for i in range(len(train_data)):
    residual += ((train_data[i][0]*a + b) - train_data[i][1])**2
print(residual)
```

```python
r_diff_a = sym.diff(residual, a)
print(r_diff_a)
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Draw the graph --> estimate the minimum is -5
xx = np.arange(-6, 6, 0.1)
yy = xx**2 + 5 * np.sin(xx)
plt.plot(xx, yy)
```

```python
# Parameters of algorithm
max_iters = 1000    # run n steps
precision = 0.000001  # fi - fi1 < epsilon 

# Given x, lr
lr        = 0.01     # learning rate
xx        = -100       

for i in range(max_iters):
    f_pre = xx**2 + 5 * np.sin(xx)
    print("Step %d: x = %f  --> f(x) = %f"%(i, xx, f_pre))
    dxy = 2*xx + 5 * np.cos(xx)
    xx = xx - lr * (dxy)
    f_cur = xx**2 + 5 * np.sin(xx)
    if abs(f_cur - f_pre) < precision: 
        break
# for
print("The local minimum occurs at x=%f with f(x) = %f."%(xx, f_cur))
```
